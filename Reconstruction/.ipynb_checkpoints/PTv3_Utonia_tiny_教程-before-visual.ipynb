# PTv3 / Utonia-tiny 入门

这个教程使用本机 `pure3d-ptv3` 环境，通过少量**模拟点云**学习如何加载权重、准备输入、提取特征，以及任务头的作用。

**操作方法：**从第一格开始，点击单元格并按 `Shift + Enter`，按顺序运行。请确认右上角内核使用本环境的 Python。

本教程不会自动对 `rot_stairs.pcd` 提取特征。最后有一个默认关闭的可选示例，只有你主动修改开关后才会处理其中的少量点。

## 1. 检查环境与导入源码

`point_transformer_v3` 已以可编辑模式安装。查看 `ptv3.__file__` 可以确认它实际引用了工作区中的哪份代码。

````python
import sys
import numpy as np
import torch
import point_transformer_v3 as ptv3

print('Python:', sys.executable)
print('PyTorch:', torch.__version__)
print('CUDA 是否可用:', torch.cuda.is_available())
print('PTv3 源码:', ptv3.__file__)
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('本教程使用设备:', device)
````

## 2. 加载 Utonia-tiny 权重

`load` 根据预设构建模型并载入权重。`warmup=False` 避免启动时运行大规模预热示例。首次加载且本机无缓存时，代码会尝试下载权重。`model.weights_info` 会显示实际权重路径。

````python
model = ptv3.load('utonia_tiny', device=device, warmup=False)
print('模型名称:', model.preset)
print('模型设备:', next(model.parameters()).device)
print('权重信息:', model.weights_info['backbone'])
````

若想手动指定权重，确保是兼容的 `utonia_tiny.pth`，然后使用：

```python
model = ptv3.load('utonia_tiny', weights='/你的路径/utonia_tiny.pth', device=device, warmup=False)
```

自动下载默认使用 Hugging Face 缓存。手动路径不必放在源码目录。

## 3. 准备模拟点云输入

模型接收 `coord`、`color` 和 `normal`，每项都是 `[N, 3]`。坐标单位为米，颜色为 RGB 0～255，法向是单位向量。Utonia 支持缺失颜色或法向时用零数组表示。

你的 MID-360 PCD 没有 RGB，不能把雷达强度直接当成 RGB。本例为学习接口，使用 128 个模拟点。

````python
rng = np.random.default_rng(7)
coord = rng.random((128, 3), dtype=np.float32)
color = np.zeros_like(coord)
normal = np.zeros_like(coord)

print('坐标:', coord.shape, '颜色:', color.shape, '法向:', normal.shape)
print('前三个点的 XYZ:')
print(coord[:3])
````

下面这幅图只展示模拟点的 XYZ，并不是建筑、模型预测或语义分割。

````python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(projection='3d')
ax.scatter(coord[:, 0], coord[:, 1], coord[:, 2], s=8)
ax.set(xlabel='X (m)', ylabel='Y (m)', zlabel='Z (m)', title='Synthetic input points')
plt.show()
````

## 4. 提取每个点的特征

`embed()` 依次执行预处理、模型前向和特征映射。Utonia-tiny 对每个输入点返回 486 个特征值。特征向量不是语义标签，也不是 trav_cost。

````python
features = model.embed(coord, color, normal)
print('输入坐标形状:', coord.shape)
print('输出特征形状:', tuple(features.shape))
print('特征设备:', features.device)
print('第一个点的前 8 维:', features[0, :8].cpu().numpy())
````

## 5. 为什么还需要任务头？

上面的 `[N, 486]` 只是特征。想输出类别，需要一个分割头。下面用随机初始化的线性层演示形状，数字没有实际语义，也不会识别真实地面或墙。

````python
demo_head = torch.nn.Linear(features.shape[1], 2).to(features.device)
demo_features = features.clone().detach()  # embed 的结果来自推理模式，复制后用于训练任务头
scores = demo_head(demo_features)
print('特征:', tuple(features.shape))
print('每点两个类别的分数:', tuple(scores.shape))
print('类别编号:', scores.argmax(dim=1)[:10].cpu().tolist())
print('以上类别由随机任务头产生，不是语义识别结果。')
````

### 可选：做一次玩具训练

下面用模拟点的 Z 坐标构造两类**玩具标签**，演示任务头怎样计算损失和更新参数。它不是你的建筑数据，也不能用来判断实际可通行性。`embed()` 使用推理模式，下面只训练任务头，不更新 Utonia。

````python
toy_labels = torch.from_numpy((coord[:, 2] > 0.5).astype(np.int64)).to(features.device)
optimizer = torch.optim.AdamW(demo_head.parameters(), lr=0.001)
optimizer.zero_grad()
scores = demo_head(demo_features)
loss = torch.nn.functional.cross_entropy(scores, toy_labels)
loss.backward()
optimizer.step()
print('本次玩具训练 loss:', round(loss.item(), 4))
````

要训练真正的行人、楼梯或栏杆识别，需要采集或准备对应语义标签。`utonia_tiny` 默认没有语义头。现成 ScanNet-20 类别也不包含专门的 person、stair、railing。`trav_cost` 则需要对应的代价标签以及机器人能力条件。

如果需要微调 Utonia 骨干，需要使用可求梯度的 `prepare -> model(point)` 训练流程。`model.embed()` 方便推理，但内部关闭梯度。

默认开关为 `False`，因此“运行全部”也不会对真实 PCD 执行模型。如需学习真实文件读取，可手动改为 `True`。这里按文件索引取样，**不是完整地图预测，也不是空间分块**。

````python
RUN_PCD_EXAMPLE = False
if RUN_PCD_EXAMPLE:
    from importlib.util import spec_from_file_location, module_from_spec
    source = Path('/home/cjh/Pure3D_ws/Reconstruction/test.py')
    spec = spec_from_file_location('pure3d_example', source)
    module = module_from_spec(spec)
    spec.loader.exec_module(module)
    pcd_xyz, pcd_normals, total = module.read_binary_pcd_sample(pcd_path, 128)
    pcd_color = np.zeros_like(pcd_xyz)
    pcd_features = model.embed(pcd_xyz, pcd_color, pcd_normals)
    print('原始点数:', total, '示例点数:', len(pcd_xyz), '特征:', tuple(pcd_features.shape))
else:
    print('真实 PCD 示例未启用。')
````
